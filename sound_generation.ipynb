# ⋆༺𓆩Xerflies! Sound Generation𓆪༻⋆


Crafting Xerfly sounds via Stable Audio Open 1.0! Prompts included in the comments, coorelated to different species of Xerflies.
https://huggingface.co/stabilityai/stable-audio-open-1.0




```python
!pip uninstall -y torchao
!pip install -q torchsde -U diffusers

import torch
import soundfile as sf
from diffusers import StableAudioPipeline
from huggingface_hub import login
from IPython.display import Audio, display

login()  # paste your HF token (skip if HF_TOKEN is set in Colab secrets)
print("installed")

pipe = StableAudioPipeline.from_pretrained(
    "stabilityai/stable-audio-open-1.0", torch_dtype=torch.float16
).to("cuda")

# Gloaming (5s): prompt = "x-ray machine beeping with ~90-120 lowpass filter"
# Silt (5s): prompt = "flapping x-ray machine deep with twinkling rocks sounds, rhythmic"
# Desert (3s): prompt = "radiographic equiptment sounds, heavy reverb, dry filter, wind, flapping"
# Juniper (4s): prompt = "bone crunch, reverb, flapping, slight pitch envelope"
# Boreal (5s): prompt = "twinkling, synth, sidechained with deep flapping sound"
# Verdant (2s): prompt = "radiographic machine sounds, loud flapping, slight reverb"
# Fulgid (3s): prompt = "short x-ray machine beep, echoing delay, wind"
# Sintered (3s): prompt = "x-ray scan sound, flapping wind, crunching leaves"

negative_prompt = "Low quality."
generator = torch.Generator("cuda").manual_seed(0)

audio = pipe(
    prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=200,
    audio_end_in_s=3.0,
    num_waveforms_per_prompt=2,
    generator=generator,
).audios

sr = pipe.vae.sampling_rate
for i, a in enumerate(audio):
    sf.write(f"output_{i}.wav", a.T.float().cpu().numpy(), sr)
    display(Audio(f"output_{i}.wav"))
```
